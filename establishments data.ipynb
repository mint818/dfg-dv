```python
import pandas as pd
```

```python
df = pd.read_csv('establishmentsdata.csv')
```

```python
df
```

```python
#health - physical & mental health 
#interested in exploring the ZIP codes and the services + resources that are offered there 
#communities that need the most help currently have less of the health services + resources 
```

```python
df1 = df[['ZIP_CODE', 'NAICS2017_LABEL']]
df1
```

```python
df2 = df1.copy() 
df2 = df2.groupby(['ZIP_CODE']).count()
df2 = df2.sort_values('NAICS2017_LABEL', ascending = True)
df2 = df2.rename(columns = {'NAICS2017_LABEL': "Total Establishments"})
df2
```

```python
#medical services (physical health) 
df3 = df1.copy() 
filter1 = df3['NAICS2017_LABEL'] == 'Offices of physicians (except mental health specialists)'
filter2 = df3['NAICS2017_LABEL'] == 'Offices of dentists'
filter3 = df3['NAICS2017_LABEL'] == 'Offices of chiropractors'
filter4 = df3['NAICS2017_LABEL'] == 'Offices of physical, occupational and speech therapists, and audiologists'
df3 = df3.where(filter1 | filter2 | filter3 | filter4)
df3 = df3.dropna()
df3 = df3.groupby(['ZIP_CODE']).count()
df3 = df3.sort_values('NAICS2017_LABEL', ascending = True)
df3 = df3.rename(columns = {'NAICS2017_LABEL': 'Medical Service Establishments'})
df3
```

```python
#recreation & entertainment (mental health)

df4 = df1.copy() 
filter1 = df4['NAICS2017_LABEL'] == 'Fitness and recreational sports centers'
filter2 = df4['NAICS2017_LABEL'] == 'Sports and recreation instruction'
filter3 = df4['NAICS2017_LABEL'] == 'Fine arts schools'
filter4 = df4['NAICS2017_LABEL'] == 'Beauty salons'
filter5 = df4['NAICS2017_LABEL'] == 'Nail salons'
filter6 = df4['NAICS2017_LABEL'] == 'Drinking places (alcoholic beverages)'
filter7 = df4['NAICS2017_LABEL'] == 'Full-service restaurants'
filter8 = df4['NAICS2017_LABEL'] == 'Limited-service restaurants'
filter9 = df4['NAICS2017_LABEL'] == 'Other personal care services'
df4 = df4.where(filter1 | filter2 | filter3 | filter4 | filter5 | filter6 | filter7 | filter8 | filter9)
df4 = df4.dropna()
df4 = df4.groupby(['ZIP_CODE']).count()
df4 = df4.sort_values('NAICS2017_LABEL', ascending = True)
df4 = df4.rename(columns = {'NAICS2017_LABEL': 'Recreation & Entertainment Establishments'})
df4
```

```python
#additional consideration 
#child-care access (more access to child care means more time for parents)

df5 = df1.copy() 
filter1 = df5['NAICS2017_LABEL'] == 'Child day care services'
df5 = df5.where(filter1)
df5 = df5.dropna()
df5 = df5.groupby(['ZIP_CODE']).count()
df5 = df5.sort_values('NAICS2017_LABEL', ascending = True)
df5 = df5.rename(columns = {'NAICS2017_LABEL': 'Child Care Services'})
df5
```
