```python
import pandas as pd
```

```python
pi_rankings = pd.read_excel("../shared_data_read_only/District_and_School_Performance_Index_Ranking.xlsx", sheet_name=2)
value_add_rankings = pd.read_excel("../shared_data_read_only/District_Value_Added_Ranking.xlsx", sheet_name = 1)
```

```python
rankings = pi_rankings
duplicateRowsName = rankings[rankings.duplicated(['LEA Name'])]
duplicateRowsIRN = rankings[rankings.duplicated(['LEA IRN'])]
duplicateRowsBoth = rankings[rankings.duplicated(['LEA IRN', 'LEA Name'])]
```

```python
len(rankings)
```

```python
combined_rankings = rankings.merge(value_add_rankings, left_on = 'LEA IRN', right_on='District IRN', how = 'inner') # Merged with Value Added File, inner because we ignored 15 entries that had NaN
combined_rankings = combined_rankings[combined_rankings['2023 PI for Ranking'] != 'NC'] # Ignore the NC since they are not calculated
combined_rankings['2023 PI for Rankings'] = combined_rankings["2023 PI for Ranking"].astype(float) 
# Removing ones that are not in both which was 15+17 and also NC which is not calculated and in respect to montessories and such
```

```python
combined_rankings[['LEA IRN', 'District IRN', 'District Name', 'LEA Name']]
```

```python
combined_rankings
```

```python
grouped_county = combined_rankings.groupby('ODE Designated County_x') # Group by county
grouped_dict_variance_county = grouped_county['2023 PI for Rankings'].var().to_dict() # Dictionary with county and variance
grouped_dict_counts_county = grouped_county.size().to_dict() # Counties with 3 of schools in it
```

```python
grouped_dict_variance_county
```

```python
# Add variance column with variance by zip code for each city
combined_rankings['Variance'] = None
for index, row in combined_rankings.iterrows():
    currCounty = row['ODE Designated County_x']
    if grouped_dict_counts_county[currCounty] >= 20:
        combined_rankings.at[index, 'Variance'] = grouped_dict_variance_county[currCounty]
    else:
        combined_rankings.at[index, 'Variance'] = 1000
```
